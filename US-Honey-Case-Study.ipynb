```python
# Create a bar-plot to display state-wise Honey production
plt.figure(figsize = (20, 8))

sns.barplot(x = 'state', y = 'production', hue = 'state', data = df)
plt.title('State-wise Honey Production')
plt.xlabel('State')
plt.ylabel('Production')
plt.xticks(rotation = 90)
plt.show()
```

```python
# Top and Bottom 5 states v/s Avg_price, Production
states_production = df.groupby('state')['production'].sum().reset_index()
states_production
```

```python
# Top 5 states honey production
Top_5_states_production = states_production.sort_values(by = 'production', ascending = False).head()
Top_5_states_production
```

```python
# Create a bar-plot to display the Top 5 states producing honey.

sns.barplot(x = 'state', y = 'production', hue = 'state', data = Top_5_states_production)
plt.title('Top 5 states producing honey')
plt.xlabel('State')
plt.ylabel('Honey Production')
plt.show()
```

```python
Bottom5_states_prod = states_production.sort_values(by='production').head()
Bottom5_states_prod
```

```python
# Bottom 5 states producing honey.

sns.barplot(x='state', y='production', hue='state', data=Bottom5_states_prod)
plt.title('Bottom 5 states producting honey')
plt.xlabel('State')
plt.ylabel('Production')
plt.show()
```

#### States V/s Average Price.

```python
states_avg_price = df.groupby('state')['average_price'].sum().reset_index()
states_avg_price
```

```python
plt.figure(figsize=(20, 8))
sns.barplot(x='state', y='average_price', hue='state', data=states_avg_price)
plt.title('State-wise Honey Price')
plt.xlabel('State')
plt.ylabel('Average Price')
plt.xticks(rotation=40)
plt.show()
```

```python
Top_5_avg = states_avg_price.sort_values(by='average_price', ascending=False).head()
Top_5_avg
```

```python
plt.pie(x='average_price', labels='state', autopct='%.1f', data=Top_5_avg)
plt.title('Top 5 states in terms of average price')
plt.show()
```

#### Year V/s Avg_price production

```python
year_avg_price = df.groupby('year')['average_price'].sum().reset_index()
year_avg_price
```

```python
plt.figure(figsize=(20, 10))
sns.set(style='darkgrid')
sns.lineplot(x='year', y='average_price', data=year_avg_price)
plt.title('Average Price Trend 1995 - 2021')
plt.xlabel('Year')
plt.ylabel('Average Price')
plt.show()
```

#### What is the change in mean average price of Honey from 1995 - 2021?

```python
mean_change_avg_df = df.groupby('year')['average_price'].mean()
mean_change_avg_df
```

```python
change_mean = mean_change_avg_df.loc[1995] - mean_change_avg_df.loc[2021]
print(change_mean)
```

#### State, Year and Production using Groupby

```python
Highest_production_state = df.groupby(['state', 'year'])['production'].sum().reset_index()
Highest_production_state
```

```python
State_2000_year = Highest_production_state.loc[(Highest_production_state['year']==2000)]
State_2000_year
```

```python
# Highest producing state in year 2000
State_2000_year.sort_values(by='production', ascending=False).head(1)
```
